#### Метрики + вспомогательные функции


```python
def calc_metrics(y_true: np.ndarray, y_pred: np.ndarray) -> tuple[float, float, float]:
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.maximum(np.asarray(y_pred, dtype=float), 0)
    # Микро-шум (floating noise): 0.0001 при факте 0 не должен давать 0% точности
    y_pred = np.where(y_pred < 1e-3, 0.0, y_pred)
    if len(y_true) == 0:
        return 0.0, 0.0, 0.0
    acc = np.empty_like(y_true)
    mask_zero = y_true == 0
    acc[mask_zero] = np.where(y_pred[mask_zero] == 0, 100.0, 0.0)
    mask_ok = ~mask_zero
    if np.any(mask_ok):
        acc[mask_ok] = np.clip(
            100.0 - np.abs(y_true[mask_ok] - y_pred[mask_ok]) / np.abs(y_true[mask_ok]) * 100.0,
            0, 100,
        )
    tochnost = round(float(np.mean(acc)), 2)
    stab = 0.0
    if tochnost > 0:
        cv = np.std(acc, ddof=0) / tochnost
        stab = round(max(0.0, (1 - cv)) * 100, 2)
    kachestvo = 0.0
    if tochnost + stab > 0:
        kachestvo = round(2 * tochnost * stab / (tochnost + stab), 2)
    return tochnost, stab, kachestvo
```

#### Per-series: `SERIES_CONFIGS`


```python
from pandas.errors import PerformanceWarning

warnings.filterwarnings("ignore", category=PerformanceWarning)

def _statsmodels_module_tag(warning) -> str:
    fn = (getattr(warning, "filename", None) or "").lower()
    if not fn:
        return ""
    if "holtwinters" in fn:
        return "holtwinters"
    if "kalman_filter" in fn:
        return "kalman_filter"
    if "sarimax" in fn:
        return "sarimax"
    if "statespace" in fn:
        return "statespace"
    if "theta" in fn and "forecasting" in fn:
        return "theta"
    return ""


def interpret_warning(msg: str, category) -> str:
    cname = getattr(category, "__name__", str(category))
    s = str(msg).lower()
    if "maximum likelihood optimization failed" in s or "convergence" in cname.lower():
        return (
            "Оптимизация MLE не сошлась: возможны короткая история, экстремальные значения ряда, "
            "плохая масштабируемость или неудачная спецификация модели."
        )
    if "invalid value encountered in scalar add" in s or "aicc" in s:
        return (
            "Численная нестабильность при расчёте AICc; часто следствие плохого фита или NaN/Inf."
        )
    if "overflow encountered in matmul" in s:
        return (
            "Переполнение при оценке ошибок (Holt-Winters): расходимость или экстремальные уровни ряда."
        )
    if "too few observations" in s and "seasonal" in s:
        return "Недостаточно наблюдений для стартовых параметров сезонной части SARIMAX."
    if "invalid value encountered in divide" in s:
        return "Некорректное деление: ноль или NaN; вырожденная дисперсия / нестабильная модель."
    if "runtime" in cname.lower():
        return "RuntimeWarning: деление на ноль, переполнение или NaN в вычислениях."
    if "sklearn" in (getattr(category, "__module__", "") or ""):
        return "Sklearn: не сходимость итераций или плохо обусловленные признаки."
    return (
        "См. текст сообщения; типично: короткий или нестабильный ряд, экстремальные значения, "
        "численная неустойчивость."
    )


def append_warning_rows(
    rows: list,
    *,
    resist_id,
    operation_id,
    model_kind: str,
    model_label: str,
    phase: str,
    n_train_points,
    caught,
):
    if not RECORD_MODEL_WARNINGS:
        return
    for w in caught:
        msg = str(w.message)
        cat = w.category
        if issubclass(cat, PerformanceWarning):
            continue
        fn = (getattr(w, "filename", None) or "").lower()
        interp = interpret_warning(msg, cat)
        if "invalid value encountered in divide" in str(msg).lower() and "kalman" in fn:
            interp = (
                "Некорректное деление в фильтре Калмана: вырожденная дисперсия / NaN в state space."
            )
        rows.append(
            {
                "resist_id": resist_id,
                "operation_id": operation_id,
                "model_kind": model_kind,
                "model_label": model_label,
                "phase": phase,
                "warning_category": getattr(cat, "__name__", str(cat)),
                "message": msg,
                "interpretation": interp,
                "n_train_points": n_train_points if model_kind == "per_series" else "",
                "statsmodels_module": _statsmodels_module_tag(w),
            }
        )


def train_ets_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    trend="add",
    damped_trend=False,
    seasonal=None,
    seasonal_periods=None,
    ets_fit_kwargs=None,
):
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if seasonal is not None and seasonal_periods is not None:
                min_hist = max(4, 2 * int(seasonal_periods))
                if len(y_hist) < min_hist:
                    raise ValueError("too short for seasonal")
            elif len(y_hist) < 4:
                raise ValueError("too short")
            if trend == "mul" and np.any(y_hist <= 0):
                raise ValueError("mul trend requires positive series")
            if seasonal == "mul" and np.any(y_hist <= 0):
                raise ValueError("mul seasonal requires positive series")
            model = ExponentialSmoothing(
                y_hist,
                trend=trend,
                damped_trend=damped_trend,
                seasonal=seasonal,
                seasonal_periods=seasonal_periods if seasonal is not None else None,
            )
            _fit_kw = {"optimized": True, "use_brute": False}
            if ets_fit_kwargs:
                _fit_kw.update(ets_fit_kwargs)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                fit = model.fit(**_fit_kw)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_fc:
                warnings.simplefilter("always")
                fc = fit.forecast(n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="forecast",
                n_train_points=n_tr,
                caught=w_fc,
            )
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_theta_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    theta_deseasonalize=True,
    theta_method="auto",
    theta_period=12,
):
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if len(y_hist) < 4:
                raise ValueError("too short")
            if theta_method == "multiplicative" and np.any(y_hist <= 0):
                raise ValueError("theta multiplicative requires positive series")
            tm = ThetaModel(
                y_hist,
                period=int(theta_period),
                deseasonalize=theta_deseasonalize,
                method=theta_method,
            )
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                res = tm.fit()
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_fc:
                warnings.simplefilter("always")
                fc = res.forecast(n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="forecast",
                n_train_points=n_tr,
                caught=w_fc,
            )
            fc = np.asarray(fc, dtype=float).ravel()
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_sarimax_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    sarimax_preset="default",
    sarimax_order=None,
    sarimax_seasonal_order=None,
    sarimax_fit_kwargs=None,
):
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if len(y_hist) < 8:
                raise ValueError("too short")
            if sarimax_order is not None and sarimax_seasonal_order is not None:
                order = tuple(sarimax_order)
                seasonal_order = tuple(sarimax_seasonal_order)
            elif len(y_hist) >= 24:
                if sarimax_preset == "airline":
                    order = (0, 1, 1)
                    seasonal_order = (0, 1, 1, 12)
                else:
                    order = (1, 1, 1)
                    seasonal_order = (1, 1, 1, 12)
            else:
                order = (1, 1, 1)
                seasonal_order = (0, 0, 0, 0)
            model = SARIMAX(
                y_hist,
                order=order,
                seasonal_order=seasonal_order,
                enforce_stationarity=False,
                enforce_invertibility=False,
            )
            _sfk = {"disp": False}
            if sarimax_fit_kwargs:
                _sfk.update(sarimax_fit_kwargs)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                fit = model.fit(**_sfk)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_fc:
                warnings.simplefilter("always")
                fc = fit.forecast(n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="forecast",
                n_train_points=n_tr,
                caught=w_fc,
            )
            fc = np.asarray(fc, dtype=float).ravel()
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_auto_arima_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    auto_arima_kwargs=None,
):
    from pmdarima import auto_arima

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    base_kw = dict(
        seasonal=True,
        m=12,
        suppress_warnings=False,
        error_action="ignore",
        stepwise=True,
        max_p=3,
        max_q=3,
        max_P=2,
        max_Q=2,
    )
    if auto_arima_kwargs:
        base_kw.update(auto_arima_kwargs)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if len(y_hist) < 8:
                raise ValueError("too short")
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                aa = auto_arima(y_hist, **base_kw)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_pred:
                warnings.simplefilter("always")
                fc = aa.predict(n_periods=n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="predict",
                n_train_points=n_tr,
                caught=w_pred,
            )
            fc = np.asarray(fc, dtype=float).ravel()
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def _sf_train_df(train_part: pd.DataFrame) -> pd.DataFrame:
    return pd.DataFrame(
        {
            "unique_id": "s0",
            "ds": pd.to_datetime(train_part["date"]),
            "y": train_part["fact_value"].values.astype(float),
        }
    )


def _sf_forecast_column(fc_df: pd.DataFrame) -> np.ndarray:
    meta = {"unique_id", "ds"}
    num_cols = [c for c in fc_df.columns if c not in meta]
    if not num_cols:
        raise ValueError("statsforecast: нет колонки прогноза")
    return fc_df[num_cols[0]].values.astype(float)


def _sf_build_model_list(sf_model_name: str, season_len: int, sf_window_size=None, tsb_kwargs=None):
    from statsforecast.models import (
        ADIDA,
        AutoARIMA,
        AutoCES,
        AutoETS,
        AutoTheta,
        CrostonSBA,
        CrostonClassic,
        DynamicOptimizedTheta,
        HistoricAverage,
        Holt,
        HoltWinters,
        IMAPA,
        MSTL,
        Naive,
        OptimizedTheta,
        RandomWalkWithDrift,
        SeasonalNaive,
        SimpleExponentialSmoothingOptimized,
        TSB,
        WindowAverage,
    )

    m = sf_model_name
    if m == "AutoETS":
        return [AutoETS(season_length=season_len)]
    if m == "AutoTheta":
        return [AutoTheta(season_length=season_len)]
    if m == "AutoCES":
        return [AutoCES(season_length=season_len)]
    if m == "AutoARIMA":
        return [AutoARIMA(season_length=season_len)]
    if m == "Naive":
        return [Naive()]
    if m == "SeasonalNaive":
        return [SeasonalNaive(season_length=season_len)]
    if m == "HistoricAverage":
        return [HistoricAverage()]
    if m == "RandomWalkWithDrift":
        return [RandomWalkWithDrift()]
    if m == "OptimizedTheta":
        return [OptimizedTheta(season_length=season_len)]
    if m == "DynamicOptimizedTheta":
        return [DynamicOptimizedTheta(season_length=season_len)]
    if m == "MSTL":
        return [MSTL(season_length=season_len)]
    if m == "Holt":
        return [Holt()]
    if m == "HoltWinters":
        return [HoltWinters(season_length=season_len)]
    if m == "SimpleExponentialSmoothingOptimized":
        return [SimpleExponentialSmoothingOptimized()]
    if m == "CrostonSBA":
        return [CrostonSBA()]
    if m == "CrostonClassic":
        return [CrostonClassic()]
    if m == "TSB":
        kw = {"alpha_d": 0.3, "alpha_p": 0.3}
        if tsb_kwargs:
            kw.update(tsb_kwargs)
        return [TSB(**kw)]
    if m == "ADIDA":
        return [ADIDA()]
    if m == "IMAPA":
        return [IMAPA()]
    if m == "WindowAverage":
        ws = int(sf_window_size) if sf_window_size is not None else 3
        ws = max(1, ws)
        return [WindowAverage(window_size=ws)]
    raise ValueError(f"Неизвестная модель StatsForecast: {sf_model_name}")


def train_statsforecast_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    sf_model_name: str,
    season_len: int,
    sf_window_size=None,
    sf_tsb_kwargs=None,
):
    from statsforecast import StatsForecast

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < 2:
                raise ValueError("too short")
            sf_df = _sf_train_df(train_part)
            models = _sf_build_model_list(sf_model_name, season_len, sf_window_size, tsb_kwargs=sf_tsb_kwargs)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                sf = StatsForecast(models=models, freq=STATSFORECAST_FREQ, n_jobs=1)
                sf.fit(sf_df)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_pred:
                warnings.simplefilter("always")
                fc_df = sf.predict(h=n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="predict",
                n_train_points=n_tr,
                caught=w_pred,
            )
            fc = _sf_forecast_column(fc_df)
            if fc.shape[0] != n_fc:
                fc = fc[:n_fc] if fc.shape[0] > n_fc else np.resize(fc, n_fc)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_statsforecast_intermittent_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    sf_model_name: str,
    season_len: int,
    sf_tsb_kwargs=None,
):
    from statsforecast import StatsForecast

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        zero_frac = float(np.mean(y_hist == 0)) if n_tr else 0.0
        use_intermittent = zero_frac >= INTERMITTENT_ZERO_MIN_FRACTION
        try:
            if n_tr < 2:
                raise ValueError("too short")
            sf_df = _sf_train_df(train_part)
            if use_intermittent:
                models = _sf_build_model_list(sf_model_name, season_len, None, tsb_kwargs=sf_tsb_kwargs)
            else:
                models = _sf_build_model_list("HistoricAverage", season_len, None)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                sf = StatsForecast(models=models, freq=STATSFORECAST_FREQ, n_jobs=1)
                sf.fit(sf_df)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_pred:
                warnings.simplefilter("always")
                fc_df = sf.predict(h=n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="predict",
                n_train_points=n_tr,
                caught=w_pred,
            )
            fc = _sf_forecast_column(fc_df)
            if fc.shape[0] != n_fc:
                fc = fc[:n_fc] if fc.shape[0] > n_fc else np.resize(fc, n_fc)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_ucm_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    ucm_level="local linear trend",
    ucm_stochastic_level=None,
    ucm_stochastic_seasonal=None,
):
    from statsmodels.tsa.statespace.structural import UnobservedComponents

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < 2 * MONTHLY_SEASON_LENGTH:
                raise ValueError("too short for seasonal UCM")
            _uk = {"level": ucm_level, "seasonal": MONTHLY_SEASON_LENGTH}
            if ucm_stochastic_level is not None:
                _uk["stochastic_level"] = ucm_stochastic_level
            if ucm_stochastic_seasonal is not None:
                _uk["stochastic_seasonal"] = ucm_stochastic_seasonal
            mod = UnobservedComponents(y_hist, **_uk)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                res = mod.fit(disp=False)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_fc:
                warnings.simplefilter("always")
                fc = res.forecast(n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="forecast",
                n_train_points=n_tr,
                caught=w_fc,
            )
            fc = np.asarray(fc, dtype=float).ravel()
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_tbats_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    tbats_use_box_cox=None,
    tbats_use_arma_errors=None,
    tbats_seasonal_periods=None,
):
    from tbats import TBATS

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < 2 * MONTHLY_SEASON_LENGTH:
                raise ValueError("too short for TBATS")
            tb_kw = {"seasonal_periods": list(tbats_seasonal_periods) if tbats_seasonal_periods is not None else [MONTHLY_SEASON_LENGTH]}
            if tbats_use_box_cox is not None:
                tb_kw["use_box_cox"] = tbats_use_box_cox
            if tbats_use_arma_errors is not None:
                tb_kw["use_arma_errors"] = tbats_use_arma_errors
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                estimator = TBATS(**tb_kw)
                fitted = estimator.fit(y_hist)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_fc:
                warnings.simplefilter("always")
                fc = fitted.forecast(steps=n_fc)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="forecast",
                n_train_points=n_tr,
                caught=w_fc,
            )
            fc = np.asarray(fc, dtype=float).ravel()
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_neuralforecast_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    nf_model_name: str,
    nf_max_steps=None,
    nf_learning_rate=None,
):
    from neuralforecast import NeuralForecast
    from neuralforecast.models import NBEATS, NHITS

    max_steps = (
        int(nf_max_steps)
        if nf_max_steps is not None
        else NEURALFORECAST_MAX_STEPS
    )
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < max(8, n_fc + 2):
                raise ValueError("too short for neural forecast")
            sf_df = _sf_train_df(train_part)
            inp = min(24, max(2, n_tr - n_fc - 1))
            inp = max(inp, 2)
            _nkw = {
                "h": n_fc,
                "input_size": inp,
                "max_steps": max_steps,
                "accelerator": "cpu",
                "devices": 1,
            }
            if nf_learning_rate is not None:
                _nkw["learning_rate"] = float(nf_learning_rate)
            if nf_model_name == "NHITS":
                model = NHITS(**_nkw)
            else:
                model = NBEATS(**_nkw)
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                nf = NeuralForecast(models=[model], freq=STATSFORECAST_FREQ)
                nf.fit(df=sf_df)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            with warnings.catch_warnings(record=True) as w_pred:
                warnings.simplefilter("always")
                fc_df = nf.predict()
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="predict",
                n_train_points=n_tr,
                caught=w_pred,
            )
            fc = _sf_forecast_column(fc_df)
            if fc.shape[0] != n_fc:
                fc = fc[:n_fc] if fc.shape[0] > n_fc else np.resize(fc, n_fc)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values



def _local_effective_max_lag(n_train: int, max_lag: int) -> int:
    if n_train < 3:
        return 1
    return max(1, min(max_lag, n_train - 2))


def _local_build_xy(y_arr: np.ndarray, max_lag: int):
    y_arr = np.asarray(y_arr, dtype=float)
    X, y_t = [], []
    for t in range(max_lag, len(y_arr)):
        X.append([y_arr[t - j] for j in range(1, max_lag + 1)])
        y_t.append(y_arr[t])
    return np.array(X, dtype=float), np.array(y_t, dtype=float)


def _local_recursive_forecast(model, scaler, y_hist: np.ndarray, n_fc: int, max_lag: int, use_scaler: bool):
    history = list(np.asarray(y_hist, dtype=float))
    fc = []
    fallback = float(np.mean(history)) if history else 0.0
    for _ in range(n_fc):
        if len(history) < max_lag:
            pred = fallback
        else:
            xrow = np.array([[history[-j] for j in range(1, max_lag + 1)]])
            if use_scaler and scaler is not None:
                xrow = scaler.transform(xrow)
            pred = float(model.predict(xrow)[0])
        fc.append(pred)
        history.append(pred)
    return np.array(fc, dtype=float)


def _glm_build_xy_count(y_arr: np.ndarray):
    y_arr = np.maximum(np.round(np.asarray(y_arr, dtype=float)), 0)
    exog = []
    endog = []
    for t in range(1, len(y_arr)):
        exog.append([1.0, float(y_arr[t - 1])])
        endog.append(float(y_arr[t]))
    return np.array(exog, dtype=float), np.array(endog, dtype=float)


def train_window_median_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    window_size: int,
):
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        history = list(y_hist)
        fc = []
        for _ in range(n_fc):
            tail = history[-window_size:] if len(history) >= window_size else history
            if len(tail) == 0:
                fc.append(0.0)
            else:
                fc.append(float(np.median(tail)))
            history.append(fc[-1])
        fc = np.array(fc, dtype=float)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_greykite_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    greykite_coverage=0.95,
    greykite_model_template=None,
    greykite_mcp_dict=None,
    greykite_extra_pred_col=False,
    greykite_forecast_one_by_one=None,
):
    from greykite.framework.templates.forecaster import Forecaster
    from greykite.framework.templates.autogen.forecast_config import (
        ForecastConfig,
        MetadataParam,
        ModelComponentsParam,
    )
    from greykite.framework.templates.model_templates import ModelTemplateEnum

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < 8:
                raise ValueError("too short for greykite")
            df_gk = pd.DataFrame({
                "ts": pd.to_datetime(train_part["date"]),
                "y": train_part["fact_value"].astype(float).values,
            })
            forecaster = Forecaster()
            try:
                tpl = ModelTemplateEnum.SILVERKITE.name
            except Exception:
                tpl = ModelTemplateEnum.AUTO.name
            metadata = MetadataParam(
                time_col="ts",
                value_col="y",
                freq=STATSFORECAST_FREQ,
            )
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                result = forecaster.run_forecast_config(
                    df=df_gk,
                    config=ForecastConfig(
                        model_template=tpl,
                        forecast_horizon=int(n_fc),
                        coverage=0.95,
                        metadata_param=metadata,
                    ),
                )
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            fdf = result.forecast.df
            fc_col = None
            for cand in ("forecast", "fcst", "yhat", "y"):
                if cand in fdf.columns:
                    fc_col = cand
                    break
            if fc_col is None:
                num_cols = [c for c in fdf.select_dtypes(include=[np.number]).columns]
                fc_col = num_cols[0] if num_cols else None
            if fc_col is None:
                raise ValueError("greykite: no forecast column")
            fc = fdf[fc_col].values.astype(float).ravel()[:n_fc]
            if fc.shape[0] < n_fc:
                fc = np.resize(fc, n_fc)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values



# Расширенный набор per-series: время прогона (runtime) и число warnings существенно растут.
SERIES_CONFIGS = {
    "ETS-add": {"fn": "ets", "trend": "add", "damped_trend": False, "seasonal": None, "seasonal_periods": None},
    "ETS-damped": {"fn": "ets", "trend": "add", "damped_trend": True, "seasonal": None, "seasonal_periods": None},
    "ETS-seasonal-add": {
        "fn": "ets",
        "trend": "add",
        "damped_trend": False,
        "seasonal": "add",
        "seasonal_periods": 12,
    },
    "ETS-seasonal-damped": {
        "fn": "ets",
        "trend": "add",
        "damped_trend": True,
        "seasonal": "add",
        "seasonal_periods": 12,
    },
    "ETS-mul-trend": {"fn": "ets", "trend": "mul", "damped_trend": False, "seasonal": None, "seasonal_periods": None},
    "ETS-seasonal-mul": {
        "fn": "ets",
        "trend": "add",
        "damped_trend": False,
        "seasonal": "mul",
        "seasonal_periods": 12,
    },
    "ETS-mul-trend-mul-seasonal": {
        "fn": "ets",
        "trend": "mul",
        "damped_trend": False,
        "seasonal": "mul",
        "seasonal_periods": 12,
    },
    "ETS-mul-trend-add-seasonal": {
        "fn": "ets",
        "trend": "mul",
        "damped_trend": False,
        "seasonal": "add",
        "seasonal_periods": 12,
    },
    "Theta-default": {"fn": "theta"},
    "Theta-no-deseason": {"fn": "theta", "theta_deseasonalize": False},
    "Theta-mult-method": {"fn": "theta", "theta_method": "multiplicative"},
    "SARIMAX-default": {"fn": "sarimax"},
    "SARIMAX-airline": {"fn": "sarimax", "sarimax_preset": "airline"},
}
if HAS_PMDARIMA:
    SERIES_CONFIGS["AutoARIMA"] = {"fn": "auto_arima"}
    SERIES_CONFIGS["AutoARIMA-nonseasonal"] = {
        "fn": "auto_arima",
        "auto_arima_kwargs": {"seasonal": False, "m": 1},
    }
    SERIES_CONFIGS["AutoARIMA-BIC"] = {
        "fn": "auto_arima",
        "auto_arima_kwargs": {"information_criterion": "bic"},
    }

if HAS_STATSFORECAST:
    _sl = MONTHLY_SEASON_LENGTH
    SERIES_CONFIGS.update(
        {
            "SF-AutoETS": {"fn": "statsforecast", "sf_model": "AutoETS", "season_len": _sl},
            "SF-AutoTheta": {"fn": "statsforecast", "sf_model": "AutoTheta", "season_len": _sl},
            "SF-AutoCES": {"fn": "statsforecast", "sf_model": "AutoCES", "season_len": _sl},
            "SF-AutoARIMA": {"fn": "statsforecast", "sf_model": "AutoARIMA", "season_len": _sl},
            "SF-Naive": {"fn": "statsforecast", "sf_model": "Naive", "season_len": _sl},
            "SF-SeasonalNaive": {"fn": "statsforecast", "sf_model": "SeasonalNaive", "season_len": _sl},
            "SF-HistoricAverage": {"fn": "statsforecast", "sf_model": "HistoricAverage", "season_len": _sl},
            "SF-RandomWalkWithDrift": {"fn": "statsforecast", "sf_model": "RandomWalkWithDrift", "season_len": _sl},
            "SF-OptimizedTheta": {"fn": "statsforecast", "sf_model": "OptimizedTheta", "season_len": _sl},
            "SF-DynamicOptimizedTheta": {"fn": "statsforecast", "sf_model": "DynamicOptimizedTheta", "season_len": _sl},
            "SF-MSTL": {"fn": "statsforecast", "sf_model": "MSTL", "season_len": _sl},
            "SF-Holt": {"fn": "statsforecast", "sf_model": "Holt", "season_len": _sl},
            "SF-HoltWinters": {"fn": "statsforecast", "sf_model": "HoltWinters", "season_len": _sl},
            "SF-SES-Optimized": {
                "fn": "statsforecast",
                "sf_model": "SimpleExponentialSmoothingOptimized",
                "season_len": _sl,
            },
            "SF-CrostonSBA": {"fn": "statsforecast_intermittent", "sf_model": "CrostonSBA", "season_len": _sl},
            "SF-CrostonClassic": {"fn": "statsforecast_intermittent", "sf_model": "CrostonClassic", "season_len": _sl},
            "SF-TSB": {"fn": "statsforecast_intermittent", "sf_model": "TSB", "season_len": _sl},
            "SF-ADIDA": {"fn": "statsforecast_intermittent", "sf_model": "ADIDA", "season_len": _sl},
            "SF-IMAPA": {"fn": "statsforecast_intermittent", "sf_model": "IMAPA", "season_len": _sl},
        }
    )


SERIES_CONFIGS["UCM-local-linear-seasonal"] = {"fn": "ucm"}
SERIES_CONFIGS["UCM-local-level-seasonal"] = {
    "fn": "ucm",
    "ucm_level": "local level",
}

if HAS_TBATS and USE_TBATS_SERIES:
    SERIES_CONFIGS["TBATS"] = {"fn": "tbats"}
    SERIES_CONFIGS["TBATS-no-boxcox"] = {"fn": "tbats", "tbats_use_box_cox": False}
    SERIES_CONFIGS["TBATS-no-arma"] = {"fn": "tbats", "tbats_use_arma_errors": False}

if HAS_NEURALFORECAST:
    SERIES_CONFIGS["NF-NHITS"] = {"fn": "neuralforecast", "nf_model": "NHITS"}
    SERIES_CONFIGS["NF-NBEATS"] = {"fn": "neuralforecast", "nf_model": "NBEATS"}
    SERIES_CONFIGS["NF-NHITS-moresteps"] = {
        "fn": "neuralforecast",
        "nf_model": "NHITS",
        "nf_max_steps": 45,
    }
    SERIES_CONFIGS["NF-NBEATS-fewersteps"] = {
        "fn": "neuralforecast",
        "nf_model": "NBEATS",
        "nf_max_steps": 15,
    }



SERIES_CONFIGS.update(
    {
        "Local-Ridge": {"fn": "local_supervised", "sk_model": "ridge"},
        "Local-Lasso": {"fn": "local_supervised", "sk_model": "lasso"},
        "Local-ElasticNet": {"fn": "local_supervised", "sk_model": "elasticnet"},
        "Local-SVR": {"fn": "local_supervised", "sk_model": "svr"},
        "Local-KNN": {"fn": "local_supervised", "sk_model": "knn"},
        "Local-RandomForest": {"fn": "local_supervised", "sk_model": "rf"},
        "GLM-Poisson": {"fn": "glm_count", "family": "poisson"},
        "GLM-NegativeBinomial": {"fn": "glm_count", "family": "negbin"},
        "ZIP": {"fn": "zip"},
        "WindowMedian-3": {"fn": "window_median", "window_size": 3},
        "WindowMedian-6": {"fn": "window_median", "window_size": 6},
    }
)
if HAS_STATSFORECAST:
    _sl2 = MONTHLY_SEASON_LENGTH
    SERIES_CONFIGS["SF-WindowAverage-3"] = {
        "fn": "statsforecast",
        "sf_model": "WindowAverage",
        "season_len": _sl2,
        "sf_window_size": 3,
    }
    SERIES_CONFIGS["SF-WindowAverage-6"] = {
        "fn": "statsforecast",
        "sf_model": "WindowAverage",
        "season_len": _sl2,
        "sf_window_size": 6,
    }

if HAS_GREYKITE and USE_GREYKITE_SERIES:
    SERIES_CONFIGS["Greykite-Silverkite"] = {"fn": "greykite"}


# -*- coding: utf-8 -*-
"""Дополнительные конфигурации по семействам fn для ноутбука 25.

Каждый ключ SERIES_CONFIGS — уникальное человекочитаемое имя (для Excel и метрик).
"""

def _register_fn_family_extras(sc: dict, ctx: dict) -> None:
    """ctx: MONTHLY_SEASON_LENGTH, HAS_PMDARIMA, HAS_STATSFORECAST, … из ячейки ноутбука.
    Trimmed: по 5 extras на семейство (из 10 в nb25).
    """
    sl = ctx["MONTHLY_SEASON_LENGTH"]
    _ets_base = {
        "fn": "ets",
        "trend": "add",
        "damped_trend": False,
        "seasonal": None,
        "seasonal_periods": None,
    }
    for name, cfg in [
        ("ETS-add-brute-search", {"ets_fit_kwargs": {"use_brute": True}}),
        ("ETS-no-trend-no-season", {"trend": None, "damped_trend": False, "seasonal": None, "seasonal_periods": None}),
        ("ETS-add-seasonal-period-6", {"trend": "add", "damped_trend": False, "seasonal": "add", "seasonal_periods": 6}),
        (
            "ETS-damped-add-seasonal-12-brute",
            {
                "trend": "add",
                "damped_trend": True,
                "seasonal": "add",
                "seasonal_periods": 12,
                "ets_fit_kwargs": {"use_brute": True},
            },
        ),
        ("ETS-add-remove-bias", {"ets_fit_kwargs": {"remove_bias": True}}),
    ]:
        sc[name] = {**_ets_base, **cfg}

    for name, cfg in [
        ("Theta-period-6-deseason-auto", {"theta_period": 6}),
        ("Theta-method-additive-deseason", {"theta_method": "additive"}),
        ("Theta-period-12-no-deseason", {"theta_period": 12, "theta_deseasonalize": False}),
        ("Theta-period-6-multiplicative-deseason", {"theta_period": 6, "theta_method": "multiplicative"}),
        ("Theta-period-24-deseason-auto", {"theta_period": 24}),
    ]:
        base = {"fn": "theta", "theta_deseasonalize": True, "theta_method": "auto", "theta_period": 12}
        base.update(cfg)
        sc[name] = base

    for name, cfg in [
        (
            "SARIMAX-order-011-season-011-m12",
            {"sarimax_order": (0, 1, 1), "sarimax_seasonal_order": (0, 1, 1, 12)},
        ),
        (
            "SARIMAX-order-111-nonseasonal",
            {"sarimax_order": (1, 1, 1), "sarimax_seasonal_order": (0, 0, 0, 0)},
        ),
        (
            "SARIMAX-order-221-nonseasonal",
            {"sarimax_order": (2, 2, 1), "sarimax_seasonal_order": (0, 0, 0, 0)},
        ),
        (
            "SARIMAX-order-011-season-111-m6",
            {"sarimax_order": (0, 1, 1), "sarimax_seasonal_order": (1, 1, 1, 6)},
        ),
        ("SARIMAX-default-fit-lbfgs", {"sarimax_fit_kwargs": {"method": "lbfgs"}}),
    ]:
        base = {"fn": "sarimax", "sarimax_preset": "default"}
        base.update(cfg)
        sc[name] = base

    if ctx["HAS_PMDARIMA"]:
        for name, cfg in [
            ("AutoARIMA-stepwise-off", {"auto_arima_kwargs": {"stepwise": False}}),
            ("AutoARIMA-ic-aicc", {"auto_arima_kwargs": {"information_criterion": "aicc"}}),
            ("AutoARIMA-max-p5-q5", {"auto_arima_kwargs": {"max_p": 5, "max_q": 5}}),
            (
                "AutoARIMA-seasonal-m12-cap-PQ1",
                {"auto_arima_kwargs": {"seasonal": True, "m": 12, "max_P": 1, "max_Q": 1}},
            ),
            ("AutoARIMA-scoring-mse", {"auto_arima_kwargs": {"scoring": "mse"}}),
        ]:
            sc[name] = {"fn": "auto_arima", **cfg}

    if ctx["HAS_STATSFORECAST"]:
        for name, w in [
            ("SF-WindowAverage-2", 2),
            ("SF-WindowAverage-4", 4),
            ("SF-WindowAverage-8", 8),
            ("SF-WindowAverage-12", 12),
            ("SF-WindowAverage-18", 18),
        ]:
            sc[name] = {
                "fn": "statsforecast",
                "sf_model": "WindowAverage",
                "season_len": sl,
                "sf_window_size": w,
            }
        for name, (ad, ap) in [
            ("SF-TSB-alpha-demand01-pool01", (0.1, 0.1)),
            ("SF-TSB-alpha-demand05-pool05", (0.5, 0.5)),
            ("SF-TSB-alpha-demand015-pool035", (0.15, 0.35)),
            ("SF-TSB-alpha-demand005-pool05", (0.05, 0.5)),
            ("SF-TSB-alpha-demand05-pool005", (0.5, 0.05)),
        ]:
            sc[name] = {
                "fn": "statsforecast_intermittent",
                "sf_model": "TSB",
                "season_len": sl,
                "sf_tsb_kwargs": {"alpha_d": ad, "alpha_p": ap},
            }


    for name, cfg in [
        ("UCM-stochastic-level-and-seasonal", {"ucm_stochastic_level": True, "ucm_stochastic_seasonal": True}),
        ("UCM-spec-random-walk", {"ucm_level": "random walk"}),
        ("UCM-spec-fixed-intercept", {"ucm_level": "fixed intercept"}),
        ("UCM-deterministic-level", {"ucm_stochastic_level": False}),
        (
            "UCM-local-level-stochastic-seasonal",
            {"ucm_level": "local level", "ucm_stochastic_seasonal": True},
        ),
    ]:
        sc[name] = {"fn": "ucm", **cfg}

    if ctx["HAS_TBATS"]:
        for name, cfg in [
            ("TBATS-seasons-6-12", {"tbats_seasonal_periods": [6, 12]}),
            ("TBATS-boxcox-and-arma-errors", {"tbats_use_box_cox": True, "tbats_use_arma_errors": True}),
            ("TBATS-seasons-3-12", {"tbats_seasonal_periods": [3, 12]}),
            ("TBATS-seasons-12-24", {"tbats_seasonal_periods": [12, 24]}),
            ("TBATS-no-arma-errors", {"tbats_use_arma_errors": False}),
        ]:
            sc[name] = {"fn": "tbats", **cfg}

    if ctx["HAS_NEURALFORECAST"]:
        for name, cfg in [
            ("NF-NHITS-steps10-lr01", {"nf_model": "NHITS", "nf_max_steps": 10, "nf_learning_rate": 0.01}),
            ("NF-NBEATS-steps60-lr001", {"nf_model": "NBEATS", "nf_max_steps": 60, "nf_learning_rate": 0.001}),
            ("NF-NHITS-steps50-lr008", {"nf_model": "NHITS", "nf_max_steps": 50, "nf_learning_rate": 0.008}),
            ("NF-NBEATS-steps40-lr02", {"nf_model": "NBEATS", "nf_max_steps": 40, "nf_learning_rate": 0.02}),
            ("NF-NHITS-steps25-lr03", {"nf_model": "NHITS", "nf_max_steps": 25, "nf_learning_rate": 0.03}),
        ]:
            sc[name] = {"fn": "neuralforecast", **cfg}

    for name, cfg in [
        ("Local-Ridge-alpha-001", {"sk_model": "ridge", "local_sk_params": {"alpha": 0.01}}),
        ("Local-Lasso-alpha-005", {"sk_model": "lasso", "local_sk_params": {"alpha": 0.05}}),
        (
            "Local-ElasticNet-alpha-02-l1ratio-02",
            {"sk_model": "elasticnet", "local_sk_params": {"alpha": 0.2, "l1_ratio": 0.2}},
        ),
        ("Local-SVR-RBF-C10-eps005", {"sk_model": "svr", "local_sk_params": {"C": 10.0, "epsilon": 0.05}}),
        ("Local-RF-depth4-estimators100", {"sk_model": "rf", "local_sk_params": {"max_depth": 4, "n_estimators": 100}}),
    ]:
        sc[name] = {"fn": "local_supervised", **cfg}

    for name, cfg in [
        (
            "GLM-Poisson-ridge-alpha-005",
            {"family": "poisson", "glm_use_regularized": True, "glm_reg_alpha": 0.05, "glm_reg_l1_wt": 0.0},
        ),
        (
            "GLM-Poisson-elastic-alpha-05-l1wt-05",
            {"family": "poisson", "glm_use_regularized": True, "glm_reg_alpha": 0.5, "glm_reg_l1_wt": 0.5},
        ),
        (
            "GLM-NegBin-ridge-alpha-01",
            {"family": "negbin", "glm_use_regularized": True, "glm_reg_alpha": 0.1, "glm_reg_l1_wt": 0.0},
        ),
        ("GLM-Poisson-wls-pinv", {"family": "poisson", "glm_fit_kwargs": {"wls_method": "pinv"}}),
        ("GLM-NegBin-maxiter-150", {"family": "negbin", "glm_fit_kwargs": {"maxiter": 150}}),
    ]:
        sc[name] = {"fn": "glm_count", **cfg}

    for name, cfg in [
        (
            "ZIP-Poisson-infl-intercept",
            {"zip_model_kind": "zip", "zip_exog_infl_mode": "intercept", "zip_fit_kwargs": {"maxiter": 200}},
        ),
        (
            "ZIP-ZINB-exog-same",
            {"zip_model_kind": "zinb", "zip_exog_infl_mode": "same", "zip_fit_kwargs": {"maxiter": 300}},
        ),
        (
            "ZIP-ZIGP-exog-same",
            {"zip_model_kind": "zigp", "zip_exog_infl_mode": "same", "zip_fit_kwargs": {"maxiter": 300}},
        ),
        (
            "ZIP-ZINB-infl-intercept",
            {"zip_model_kind": "zinb", "zip_exog_infl_mode": "intercept", "zip_fit_kwargs": {"maxiter": 300}},
        ),
        (
            "ZIP-Poisson-bfgs",
            {"zip_model_kind": "zip", "zip_exog_infl_mode": "same", "zip_fit_kwargs": {"method": "bfgs", "maxiter": 200}},
        ),
    ]:
        sc[name] = {"fn": "zip", **cfg}

    for name, ws in [
        ("WindowMedian-2", 2),
        ("WindowMedian-4", 4),
        ("WindowMedian-8", 8),
        ("WindowMedian-12", 12),
        ("WindowMedian-18", 18),
    ]:
        sc[name] = {"fn": "window_median", "window_size": ws}

    if ctx["HAS_GREYKITE"]:

        def _gk_mcp(fit_algorithm: str):
            return {"custom": {"fit_algorithm_dict": {"fit_algorithm": fit_algorithm}}}

        for name, cfg in [
            ("Greykite-template-auto", {"greykite_model_template": "AUTO"}),
            ("Greykite-silverkite-ridge", {"greykite_mcp_dict": _gk_mcp("ridge")}),
            ("Greykite-silverkite-gb", {"greykite_mcp_dict": _gk_mcp("gradient_boosting")}),
            ("Greykite-silverkite-lasso", {"greykite_mcp_dict": _gk_mcp("lasso")}),
            ("Greykite-forecast-one-by-one", {"greykite_forecast_one_by_one": 1}),
        ]:
            sc[name] = {"fn": "greykite", "greykite_coverage": 0.95, **cfg}

_register_fn_family_extras(
    SERIES_CONFIGS,
    {
        "MONTHLY_SEASON_LENGTH": MONTHLY_SEASON_LENGTH,
        "HAS_PMDARIMA": HAS_PMDARIMA,
        "HAS_STATSFORECAST": HAS_STATSFORECAST,
        "HAS_PROPHET": False,
        "HAS_TBATS": HAS_TBATS and USE_TBATS_SERIES,
        "HAS_NEURALFORECAST": HAS_NEURALFORECAST,
        "HAS_GREYKITE": HAS_GREYKITE and USE_GREYKITE_SERIES,
    },
)


def _build_model_configs() -> dict:
    """Фабрика: каждый вызов создаёт свежие (не обученные) экземпляры моделей."""
    return {
        "LinearRegression": LinearRegression(),
        "DT-shallow": DecisionTreeRegressor(max_depth=6, min_samples_leaf=5, random_state=42),
        "DT-deep": DecisionTreeRegressor(max_depth=12, min_samples_leaf=2, random_state=42),
        "DT-mid": DecisionTreeRegressor(max_depth=8, min_samples_leaf=4, random_state=42),
        "DT-conservative": DecisionTreeRegressor(
            max_depth=5, min_samples_leaf=10, min_samples_split=15, random_state=42,
        ),
        "DT-ccp": DecisionTreeRegressor(
            max_depth=15, min_samples_leaf=2, ccp_alpha=0.02, random_state=42,
        ),
        "LGBM-default": LGBMRegressor(
            n_estimators=500, max_depth=6, learning_rate=0.1,
            verbose=-1, random_state=42,
        ),
        "LGBM-tuned": LGBMRegressor(
            n_estimators=800, max_depth=8, learning_rate=0.05,
            num_leaves=63, min_child_samples=10, verbose=-1, random_state=42,
        ),
        "XGB-default": XGBRegressor(
            n_estimators=500, max_depth=6, learning_rate=0.1,
            verbosity=0, random_state=42,
        ),
        "XGB-regularized": XGBRegressor(
            n_estimators=800, max_depth=8, learning_rate=0.05,
            reg_alpha=0.1, reg_lambda=5, verbosity=0, random_state=42,
        ),
        "RF-default": RandomForestRegressor(
            n_estimators=300, max_depth=12, min_samples_leaf=2,
            random_state=42, n_jobs=-1,
        ),
        "ET-default": ExtraTreesRegressor(
            n_estimators=300, max_depth=12, min_samples_leaf=2,
            random_state=42, n_jobs=-1,
        ),
        "HistGBM-default": HistGradientBoostingRegressor(
            max_iter=500, max_depth=6, learning_rate=0.1,
            random_state=42,
        ),
        "GBR-default": GradientBoostingRegressor(
            n_estimators=300, max_depth=6, learning_rate=0.1,
            random_state=42,
        ),
    }


# --- Notebook 32: расширения feature_mode (engineered vs raw K-lags) для local_supervised / GLM / ZIP ---

def _history_to_engineered_row(history: list, train_med: float) -> np.ndarray:
    """Вектор признаков как LAG_FEATURES: lag_1..lag_6, lag_12, rolling_mean_3."""
    h = [float(x) for x in history]

    def lg(k: int) -> float:
        return float(h[-k]) if len(h) >= k else float(train_med)

    lags = [lg(i) for i in range(1, 7)]
    l12 = lg(12)
    tail3 = h[-3:] if len(h) >= 3 else h
    rm3 = float(np.mean(tail3)) if tail3 else 0.0
    return np.array(lags + [l12, rm3], dtype=float)


def train_local_supervised_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    sk_model: str,
    max_lag: int = None,
    local_sk_params=None,
    feature_mode: str = "raw",
    lag_feature_cols=None,
):
    """feature_mode: 'raw' — окно из y (как в 30); 'engineered' — LAG_FEATURES из df + рекурсия по истории."""
    from sklearn.linear_model import Ridge, Lasso, ElasticNet
    from sklearn.svm import SVR
    from sklearn.neighbors import KNeighborsRegressor
    from sklearn.ensemble import RandomForestRegressor
    from sklearn.preprocessing import StandardScaler

    if lag_feature_cols is None:
        lag_feature_cols = list(LAG_FEATURES)
    if max_lag is None:
        max_lag = LOCAL_MAX_LAG
    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        train_med = float(np.median(y_hist)) if len(y_hist) else 0.0
        try:
            if feature_mode == "engineered":
                X_tr = train_part[lag_feature_cols].values.astype(float)
                y_tr = y_hist.copy()
                if X_tr.shape[0] < 3:
                    raise ValueError("too few rows")
                L_eff = X_tr.shape[1]
                use_scaler = sk_model in ("svr", "knn")
                scaler = StandardScaler() if use_scaler else None
                X_fit = scaler.fit_transform(X_tr) if use_scaler else X_tr
                _p = dict(local_sk_params or {})
                if sk_model == "ridge":
                    model = Ridge(**{**{"random_state": 42}, **_p})
                elif sk_model == "lasso":
                    model = Lasso(**{**{"random_state": 42, "max_iter": 5000}, **_p})
                elif sk_model == "elasticnet":
                    model = ElasticNet(**{**{"random_state": 42, "max_iter": 5000}, **_p})
                elif sk_model == "svr":
                    model = SVR(kernel="rbf", **_p)
                elif sk_model == "knn":
                    model = KNeighborsRegressor(**{**{"n_jobs": -1}, **_p})
                elif sk_model == "rf":
                    model = RandomForestRegressor(**{**{"random_state": 42, "n_jobs": -1}, **_p})
                else:
                    raise ValueError(sk_model)
                with warnings.catch_warnings(record=True) as w_fit:
                    warnings.simplefilter("always")
                    model.fit(X_fit, y_tr)
                append_warning_rows(
                    warning_rows,
                    resist_id=resist_id,
                    operation_id=op_id,
                    model_kind="per_series",
                    model_label=series_config_name,
                    phase="fit",
                    n_train_points=n_tr,
                    caught=w_fit,
                )
                history = list(y_hist)
                fc = []
                for _ in range(n_fc):
                    xrow = _history_to_engineered_row(history, train_med).reshape(1, -1)
                    if xrow.shape[1] != L_eff:
                        raise ValueError("lag dim mismatch")
                    x_pred = scaler.transform(xrow) if use_scaler else xrow
                    pred = float(model.predict(x_pred)[0])
                    fc.append(pred)
                    history.append(pred)
                fc = np.array(fc, dtype=float)
            else:
                L = _local_effective_max_lag(n_tr, max_lag)
                use_scaler = sk_model in ("svr", "knn")
                X, y_tgt = _local_build_xy(y_hist, L)
                if X.shape[0] < 2:
                    raise ValueError("too few supervised rows")
                scaler = StandardScaler() if use_scaler else None
                X_fit = scaler.fit_transform(X) if use_scaler else X
                _p = dict(local_sk_params or {})
                if sk_model == "ridge":
                    model = Ridge(**{**{"random_state": 42}, **_p})
                elif sk_model == "lasso":
                    model = Lasso(**{**{"random_state": 42, "max_iter": 5000}, **_p})
                elif sk_model == "elasticnet":
                    model = ElasticNet(**{**{"random_state": 42, "max_iter": 5000}, **_p})
                elif sk_model == "svr":
                    model = SVR(kernel="rbf", **_p)
                elif sk_model == "knn":
                    model = KNeighborsRegressor(**{**{"n_jobs": -1}, **_p})
                elif sk_model == "rf":
                    model = RandomForestRegressor(**{**{"random_state": 42, "n_jobs": -1}, **_p})
                else:
                    raise ValueError(sk_model)
                with warnings.catch_warnings(record=True) as w_fit:
                    warnings.simplefilter("always")
                    model.fit(X_fit, y_tgt)
                append_warning_rows(
                    warning_rows,
                    resist_id=resist_id,
                    operation_id=op_id,
                    model_kind="per_series",
                    model_label=series_config_name,
                    phase="fit",
                    n_train_points=n_tr,
                    caught=w_fit,
                )
                fc = _local_recursive_forecast(model, scaler, y_hist, n_fc, L, use_scaler)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def _glm_build_xy_count_k(y_arr: np.ndarray, n_lags: int):
    y_arr = np.asarray(y_arr, dtype=float)
    K = max(1, int(n_lags))
    exog, endog = [], []
    for t in range(K, len(y_arr)):
        row = [1.0] + [float(y_arr[t - j]) for j in range(1, K + 1)]
        exog.append(row)
        endog.append(float(y_arr[t]))
    return np.array(exog, dtype=float), np.array(endog, dtype=float)


def train_glm_count_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    family_name: str,
    glm_fit_kwargs=None,
    glm_use_regularized: bool = False,
    glm_reg_alpha: float = 0.1,
    glm_reg_l1_wt: float = 0.0,
    glm_n_lags: int = 1,
):
    from statsmodels.genmod.generalized_linear_model import GLM
    from statsmodels.genmod import families

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    fam = families.Poisson() if family_name == "poisson" else families.NegativeBinomial()
    K = max(1, int(glm_n_lags))
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < K + 3:
                raise ValueError("too short")
            exog_tr, endog_tr = _glm_build_xy_count_k(y_hist, K)
            if exog_tr.shape[0] < 3:
                raise ValueError("too few glm rows")
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                glm = GLM(endog_tr, exog_tr, family=fam)
                if glm_use_regularized:
                    res = glm.fit_regularized(
                        alpha=float(glm_reg_alpha),
                        L1_wt=float(glm_reg_l1_wt),
                        maxiter=int((glm_fit_kwargs or {}).get("maxiter", 100)),
                    )
                else:
                    _gkw = {}
                    if glm_fit_kwargs:
                        _gkw.update(glm_fit_kwargs)
                    res = glm.fit(**_gkw)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            history = list(np.maximum(np.round(y_hist), 0).astype(float))
            fc = []
            for _ in range(n_fc):
                x1 = np.array([[1.0] + [history[-j] for j in range(1, K + 1)]])
                mu = float(np.squeeze(res.predict(x1)))
                pred = max(0.0, mu)
                fc.append(pred)
                history.append(float(max(0, round(pred))))
            fc = np.array(fc, dtype=float)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def train_zip_per_series(
    df_full,
    mask_train,
    mask_test,
    resist_id,
    series_config_name,
    warning_rows,
    *,
    zip_fit_kwargs=None,
    zip_model_kind="zip",
    zip_exog_infl_mode="same",
    zip_n_lags: int = 1,
):
    from statsmodels.discrete.count_model import (
        ZeroInflatedPoisson,
        ZeroInflatedNegativeBinomialP,
        ZeroInflatedGeneralizedPoisson,
    )

    test_idx = df_full.loc[mask_test].index
    preds = pd.Series(0.0, index=test_idx)
    K = max(1, int(zip_n_lags))
    for op_id, grp in df_full.groupby("operation_id"):
        train_part = grp[mask_train.reindex(grp.index, fill_value=False)].sort_values("date")
        test_part = grp[mask_test.reindex(grp.index, fill_value=False)].sort_values("date")
        if test_part.empty:
            continue
        y_hist = train_part["fact_value"].values.astype(float)
        n_fc = len(test_part)
        n_tr = len(y_hist)
        try:
            if n_tr < K + 5:
                raise ValueError("too short for zip")
            y_int = np.maximum(np.round(y_hist), 0).astype(int)
            exog_rows = []
            endog_tr = []
            for t in range(K, len(y_int)):
                exog_rows.append([1.0] + [float(y_int[t - j]) for j in range(1, K + 1)])
                endog_tr.append(float(y_int[t]))
            exog_tr = np.array(exog_rows, dtype=float)
            endog_tr = np.array(endog_tr, dtype=float)
            if exog_tr.shape[0] < 4:
                raise ValueError("too few zip rows")
            with warnings.catch_warnings(record=True) as w_fit:
                warnings.simplefilter("always")
                exog_infl_tr = exog_tr if zip_exog_infl_mode == "same" else np.ones((exog_tr.shape[0], 1))
                if zip_model_kind == "zinb":
                    ModelCls = ZeroInflatedNegativeBinomialP
                elif zip_model_kind == "zigp":
                    ModelCls = ZeroInflatedGeneralizedPoisson
                else:
                    ModelCls = ZeroInflatedPoisson
                z = ModelCls(endog_tr, exog_tr, exog_infl=exog_infl_tr)
                _zkw = {"disp": False, "maxiter": 200}
                if zip_fit_kwargs:
                    _zkw.update(zip_fit_kwargs)
                zres = z.fit(**_zkw)
            append_warning_rows(
                warning_rows,
                resist_id=resist_id,
                operation_id=op_id,
                model_kind="per_series",
                model_label=series_config_name,
                phase="fit",
                n_train_points=n_tr,
                caught=w_fit,
            )
            history = list(y_int.astype(float))
            fc = []
            for _ in range(n_fc):
                x1 = np.array([[1.0] + [history[-j] for j in range(1, K + 1)]])
                infl1 = x1 if zip_exog_infl_mode == "same" else np.array([[1.0]])
                mu = float(np.squeeze(zres.predict(exog=x1, exog_infl=infl1)))
                pred = max(0.0, mu)
                fc.append(pred)
                history.append(float(max(0, round(pred))))
            fc = np.array(fc, dtype=float)
        except Exception:
            fc = np.full(n_fc, np.mean(y_hist) if len(y_hist) > 0 else 0)
        preds.loc[test_part.index] = np.maximum(fc, 0)
    return preds.values


def _prepare_ops_df_for_rid(df_model39, rid, TEST_YEAR, TEST_MONTHS):
    """Лаги (lags) и маски train/test как в ноутбуке 30."""
    df = df_model39[df_model39["resist_id"] == rid].copy()
    df["op_code"] = df["operation_id"].astype("category").cat.codes
    group = df.groupby("operation_id")
    df["lag_1"] = group["fact_value"].shift(1)
    for _k in range(2, 7):
        df[f"lag_{_k}"] = group["fact_value"].shift(_k)
    df["lag_12"] = group["fact_value"].shift(12)
    df["rolling_mean_3"] = group["fact_value"].transform(
        lambda s: s.shift(1).rolling(window=3, min_periods=1).mean()
    )
    mask_test = (df["year"] == TEST_YEAR) & (df["month"].isin(TEST_MONTHS))
    mask_train = ~mask_test
    half_year_lag_cols = [f"lag_{i}" for i in range(1, 7)]
    train_med = df.loc[mask_train].groupby("operation_id")["fact_value"].median()
    for _col in half_year_lag_cols:
        df[_col] = df[_col].fillna(df["operation_id"].map(train_med)).fillna(0.0)
    num_cols = df.select_dtypes(include=["number"]).columns
    df[num_cols] = df[num_cols].fillna(0)
    df = df.infer_objects(copy=False)
    if df.isnull().sum().sum() > 0:
        df = df.fillna(0)
    return df, mask_train, mask_test


def _per_series_y_pred_from_cfg(
    df, mask_train, mask_test, rid, cfg: dict, feature_set: str, raw_max_lag: int | None, warning_rows: list,
):
    """Диспетчер (dispatcher) train_* для одного resist_id. Возвращает y_pred или None."""
    cfg = dict(cfg)
    fn = cfg["fn"]
    name = cfg.get("_label", "anon")
    if fn == "ets":
        return train_ets_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            trend=cfg["trend"], damped_trend=cfg["damped_trend"],
            seasonal=cfg.get("seasonal"), seasonal_periods=cfg.get("seasonal_periods"),
            ets_fit_kwargs=cfg.get("ets_fit_kwargs"),
        )
    if fn == "theta":
        return train_theta_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            theta_deseasonalize=cfg.get("theta_deseasonalize", True),
            theta_method=cfg.get("theta_method", "auto"),
            theta_period=cfg.get("theta_period", 12),
        )
    if fn == "sarimax":
        return train_sarimax_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            sarimax_preset=cfg.get("sarimax_preset", "default"),
            sarimax_order=cfg.get("sarimax_order"),
            sarimax_seasonal_order=cfg.get("sarimax_seasonal_order"),
            sarimax_fit_kwargs=cfg.get("sarimax_fit_kwargs"),
        )
    if fn == "auto_arima":
        return train_auto_arima_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            auto_arima_kwargs=cfg.get("auto_arima_kwargs"),
        )
    if fn == "statsforecast":
        return train_statsforecast_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            sf_model_name=cfg["sf_model"],
            season_len=cfg["season_len"],
            sf_window_size=cfg.get("sf_window_size"),
            sf_tsb_kwargs=cfg.get("sf_tsb_kwargs"),
        )
    if fn == "statsforecast_intermittent":
        return train_statsforecast_intermittent_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            sf_model_name=cfg["sf_model"],
            season_len=cfg["season_len"],
            sf_tsb_kwargs=cfg.get("sf_tsb_kwargs"),
        )
    if fn == "ucm":
        return train_ucm_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            ucm_level=cfg.get("ucm_level", "local linear trend"),
            ucm_stochastic_level=cfg.get("ucm_stochastic_level"),
            ucm_stochastic_seasonal=cfg.get("ucm_stochastic_seasonal"),
        )
    if fn == "tbats":
        return train_tbats_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            tbats_use_box_cox=cfg.get("tbats_use_box_cox"),
            tbats_use_arma_errors=cfg.get("tbats_use_arma_errors"),
            tbats_seasonal_periods=cfg.get("tbats_seasonal_periods"),
        )
    if fn == "neuralforecast":
        return train_neuralforecast_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            nf_model_name=cfg["nf_model"],
            nf_max_steps=cfg.get("nf_max_steps"),
            nf_learning_rate=cfg.get("nf_learning_rate"),
        )
    if fn == "local_supervised":
        k = int(raw_max_lag or LOCAL_MAX_LAG)
        if feature_set == "engineered_features":
            return train_local_supervised_per_series(
                df, mask_train, mask_test, rid, name, warning_rows,
                sk_model=cfg["sk_model"],
                local_sk_params=cfg.get("local_sk_params"),
                feature_mode="engineered",
            )
        return train_local_supervised_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            sk_model=cfg["sk_model"],
            local_sk_params=cfg.get("local_sk_params"),
            max_lag=k,
            feature_mode="raw",
        )
    if fn == "glm_count":
        nk = int(cfg.get("glm_n_lags", 1))
        if feature_set == "raw_sequence_features":
            nk = max(nk, int(raw_max_lag or nk))
        else:
            nk = min(max(nk, 1), 6)
        return train_glm_count_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            family_name=cfg["family"],
            glm_fit_kwargs=cfg.get("glm_fit_kwargs"),
            glm_use_regularized=cfg.get("glm_use_regularized", False),
            glm_reg_alpha=cfg.get("glm_reg_alpha", 0.1),
            glm_reg_l1_wt=cfg.get("glm_reg_l1_wt", 0.0),
            glm_n_lags=nk,
        )
    if fn == "zip":
        nk = int(cfg.get("zip_n_lags", 1))
        if feature_set == "raw_sequence_features":
            nk = max(nk, int(raw_max_lag or nk))
        else:
            nk = min(max(nk, 1), 6)
        return train_zip_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            zip_fit_kwargs=cfg.get("zip_fit_kwargs"),
            zip_model_kind=cfg.get("zip_model_kind", "zip"),
            zip_exog_infl_mode=cfg.get("zip_exog_infl_mode", "same"),
            zip_n_lags=nk,
        )
    if fn == "window_median":
        return train_window_median_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            window_size=int(cfg["window_size"]),
        )
    if fn == "greykite":
        return train_greykite_per_series(
            df, mask_train, mask_test, rid, name, warning_rows,
            greykite_coverage=cfg.get("greykite_coverage", 0.95),
            greykite_model_template=cfg.get("greykite_model_template"),
            greykite_mcp_dict=cfg.get("greykite_mcp_dict"),
            greykite_extra_pred_col=cfg.get("greykite_extra_pred_col", False),
            greykite_forecast_one_by_one=cfg.get("greykite_forecast_one_by_one"),
        )
    return None


def run_forecast_one_config(
    df_model39,
    RESIST_IDS,
    TEST_YEAR,
    TEST_MONTHS,
    cfg: dict,
    feature_set: str,
    raw_max_lag: int | None,
    warning_rows: list,
) -> tuple[float, float, float, float, int]:
    """Один полный прогон по всем RESIST_IDS; возвращает (mean_quality, mean_acc, mean_stab, total_sec, n_points)."""
    t_all0 = time.perf_counter()
    qs, ts, ss = [], [], []
    for rid in RESIST_IDS:
        df, mask_train, mask_test = _prepare_ops_df_for_rid(df_model39, rid, TEST_YEAR, TEST_MONTHS)
        if mask_train.sum() == 0 or int(mask_test.sum()) == 0:
            continue
        y_test = df.loc[mask_test, "fact_value"]
        y_pred = _per_series_y_pred_from_cfg(
            df, mask_train, mask_test, rid, cfg, feature_set, raw_max_lag, warning_rows,
        )
        if y_pred is None:
            continue
        tochnost, stab, kachestvo = calc_metrics(y_test.values, y_pred)
        qs.append(kachestvo)
        ts.append(tochnost)
        ss.append(stab)
    elapsed = time.perf_counter() - t_all0
    if not qs:
        return 0.0, 0.0, 0.0, elapsed, 0
    return float(np.mean(qs)), float(np.mean(ts)), float(np.mean(ss)), elapsed, len(qs)


def run_forecast_collect_detail(
    df_model39,
    RESIST_IDS,
    TEST_YEAR,
    TEST_MONTHS,
    cfg: dict,
    feature_set: str,
    raw_max_lag: int | None,
    combo_label: str,
    warning_rows: list | None = None,
):
    """Таблицы «Данные теста» и «По операциям» как в ноутбуке 30 — для одного выбранного конфига (config)."""
    wr = warning_rows if warning_rows is not None else []
    per_rid = {}
    for rid in RESIST_IDS:
        df, mask_train, mask_test = _prepare_ops_df_for_rid(df_model39, rid, TEST_YEAR, TEST_MONTHS)
        if mask_train.sum() == 0 or int(mask_test.sum()) == 0:
            continue
        y_pred = _per_series_y_pred_from_cfg(
            df, mask_train, mask_test, rid, cfg, feature_set, raw_max_lag, wr,
        )
        if y_pred is None:
            continue
        df_test = df.loc[mask_test].copy()
        df_test["predict_new"] = y_pred
        rows_op = []
        for (r_id, op_id), grp in df_test.groupby(["resist_id", "operation_id"]):
            t1, s1, q1 = calc_metrics(grp["fact_value"].values, grp["fact_predict"].values)
            t2, s2, q2 = calc_metrics(grp["fact_value"].values, grp["predict_new"].values)
            rows_op.append({
                "resist_id": r_id,
                "operation_id": op_id,
                "Точность (fact_predict)": t1,
                "Стабильность (fact_predict)": s1,
                "Качество (fact_predict)": q1,
                f"Точность ({combo_label})": t2,
                f"Стабильность ({combo_label})": s2,
                f"Качество ({combo_label})": q2,
            })
        df_metrics_op = pd.DataFrame(rows_op)
        df_detail = df_test[["resist_id", "operation_id", "year", "month",
                             "fact_value", "fact_predict", "predict_new"]].copy()
        df_detail = df_detail.sort_values(["resist_id", "operation_id", "year", "month"]).reset_index(drop=True)
        df_detail = df_detail.rename(columns={
            "fact_value": "fact_value (факт)",
            "fact_predict": "fact_predict (текущий)",
            "predict_new": f"{combo_label} (новый)",
        })
        per_rid[rid] = {"df_detail": df_detail, "df_metrics_op": df_metrics_op}
    return per_rid

```

```python
# --- Random search: пространства параметров и сэмплирование ---
import random
rng = random.Random(SEARCH_SEED)

def _cfg_label(fn: str, d: dict) -> str:
    slim = {k: v for k, v in d.items() if not k.startswith("_")}
    return f"{fn}|{json.dumps(slim, sort_keys=True, default=str)[:180]}"


def sample_config_for_fn(fn: str) -> dict | None:
    """Возвращает dict с ключом 'fn' и полями, совместимыми с run_forecast_one_config."""
    sl = MONTHLY_SEASON_LENGTH
    if fn == "ets":
        trend = rng.choice(["add", "mul"])
        damped = rng.choice([True, False])
        sea = rng.choice([None, "add"])
        sp = 12 if sea else None
        return {"fn": "ets", "trend": trend, "damped_trend": damped, "seasonal": sea, "seasonal_periods": sp,
                "_label": _cfg_label("ets", {"t": trend, "d": damped, "s": sea})}
    if fn == "theta":
        return {"fn": "theta", "theta_deseasonalize": rng.choice([True, False]),
                "theta_method": rng.choice(["auto", "additive"]),
                "theta_period": rng.choice([12, 6]),
                "_label": _cfg_label("theta", {})}
    if fn == "sarimax":
        presets = [
            ((1, 1, 1), (1, 1, 1, 12)),
            ((0, 1, 1), (0, 1, 1, 12)),
            ((1, 1, 1), (0, 0, 0, 0)),
        ]
        order, seas = rng.choice(presets)
        return {"fn": "sarimax", "sarimax_preset": "default", "sarimax_order": order,
                "sarimax_seasonal_order": seas, "_label": _cfg_label("sarimax", {"o": order})}
    if fn == "auto_arima":
        if not HAS_PMDARIMA:
            return None
        return {"fn": "auto_arima", "auto_arima_kwargs": {
            "stepwise": rng.choice([True, False]),
            "information_criterion": rng.choice(["aicc", "bic"]),
        }, "_label": _cfg_label("auto_arima", {})}
    if fn == "statsforecast":
        if not HAS_STATSFORECAST:
            return None
        mname = rng.choice(["AutoETS", "AutoTheta", "SeasonalNaive", "WindowAverage"])
        w = rng.choice([2, 4, 8, 12])
        cfg = {"fn": "statsforecast", "sf_model": mname, "season_len": sl, "_label": _cfg_label("sf", {"m": mname})}
        if mname == "WindowAverage":
            cfg["sf_window_size"] = w
        return cfg
    if fn == "statsforecast_intermittent":
        if not HAS_STATSFORECAST:
            return None
        return {"fn": "statsforecast_intermittent", "sf_model": rng.choice(["CrostonSBA", "TSB", "ADIDA"]),
                "season_len": sl, "_label": _cfg_label("sfi", {})}
    if fn == "ucm":
        return {"fn": "ucm", "ucm_level": rng.choice(["local linear trend", "local level", "random walk"]),
                "ucm_stochastic_level": rng.choice([None, True, False]),
                "_label": _cfg_label("ucm", {})}
    if fn == "tbats":
        if not (HAS_TBATS and USE_TBATS_SERIES):
            return None
        return {"fn": "tbats", "tbats_use_box_cox": rng.choice([True, False]),
                "tbats_use_arma_errors": rng.choice([True, False]),
                "tbats_seasonal_periods": rng.choice([[12], [6, 12], [3, 12]]),
                "_label": _cfg_label("tbats", {})}
    if fn == "neuralforecast":
        if not HAS_NEURALFORECAST:
            return None
        return {"fn": "neuralforecast", "nf_model": rng.choice(["NHITS", "NBEATS"]),
                "nf_max_steps": rng.choice([15, 25, 40]),
                "nf_learning_rate": rng.choice([0.001, 0.01, 0.02]),
                "_label": _cfg_label("nf", {})}
    if fn == "local_supervised":
        sk = rng.choice(["ridge", "lasso", "elasticnet", "rf"])
        pr = {"ridge": {"alpha": rng.choice([0.01, 0.1, 1.0])},
              "lasso": {"alpha": rng.choice([0.01, 0.05, 0.1])},
              "elasticnet": {"alpha": rng.choice([0.1, 0.2]), "l1_ratio": rng.choice([0.2, 0.5])},
              "rf": {"max_depth": rng.choice([4, 8]), "n_estimators": rng.choice([50, 100])}}[sk]
        return {"fn": "local_supervised", "sk_model": sk, "local_sk_params": pr,
                "_label": _cfg_label("loc", {"sk": sk})}
    if fn == "glm_count":
        return {"fn": "glm_count", "family": rng.choice(["poisson", "negbin"]),
                "glm_use_regularized": rng.choice([True, False]),
                "glm_reg_alpha": rng.choice([0.01, 0.05, 0.1]),
                "glm_n_lags": rng.choice([1, 2, 3]),
                "_label": _cfg_label("glm", {})}
    if fn == "zip":
        return {"fn": "zip", "zip_model_kind": rng.choice(["zip", "zinb"]),
                "zip_exog_infl_mode": rng.choice(["same", "intercept"]),
                "zip_n_lags": rng.choice([1, 2, 3]),
                "_label": _cfg_label("zip", {})}
    if fn == "window_median":
        return {"fn": "window_median", "window_size": rng.choice([3, 6, 12, 18]),
                "_label": _cfg_label("wmed", {})}
    if fn == "greykite":
        if not (HAS_GREYKITE and USE_GREYKITE_SERIES):
            return None
        return {"fn": "greykite", "greykite_coverage": 0.95,
                "greykite_model_template": rng.choice([None, "AUTO"]),
                "_label": _cfg_label("gk", {})}
    return None


def refine_config(base: dict) -> dict:
    c = json.loads(json.dumps(base))
    fn = c["fn"]
    if fn == "window_median":
        c["window_size"] = int(max(2, c.get("window_size", 6) + rng.choice([-1, 0, 1])))
    if fn == "ets" and c.get("seasonal") is not None:
        c["damped_trend"] = not c.get("damped_trend", False)
    if fn == "local_supervised" and c.get("local_sk_params"):
        a = c["local_sk_params"].get("alpha", 0.1)
        c["local_sk_params"]["alpha"] = float(max(1e-4, a * rng.choice([0.5, 1.0, 2.0])))
    c["_label"] = c.get("_label", "") + ":ref"
    return c


FN_ORDER = [
    "ets", "theta", "sarimax", "auto_arima", "statsforecast", "statsforecast_intermittent",
    "ucm", "tbats", "neuralforecast", "local_supervised", "glm_count", "zip", "window_median", "greykite",
]

attempt_log: list[dict] = []
family_summaries: list[dict] = []


def run_search_univariate(fn: str):
    """Один цикл на fn: общая ранняя остановка; в лог — по 2 строки на попытку (engineered + raw)."""
    if sample_config_for_fn(fn) is None:
        family_summaries.append({
            "model_family": fn,
            "feature_set": "both_logged",
            "total_attempts": 0,
            "total_forecasts": 0,
            "best_params": "",
            "best_quality": -1.0,
            "reason_for_stop": "skipped_unavailable",
        })
        return
    best_q = -1.0
    no_gain = 0
    attempts = 0
    forecasts = 0
    best_params = None
    reason = ""
    trial_qualities: list[tuple[float, dict]] = []
    refinement_queue: list[dict] = []
    refine_seeded = False
    while attempts < MAX_ATTEMPTS_PER_MODEL:
        if (not refine_seeded) and USE_REFINEMENT and attempts >= MAX_ATTEMPTS_PER_MODEL // 2 and trial_qualities:
            trial_qualities.sort(key=lambda x: -x[0])
            for _, cfg in trial_qualities[:3]:
                refinement_queue.extend([refine_config(cfg), refine_config(cfg)])
            refine_seeded = True
        if refinement_queue:
            cfg = refinement_queue.pop(0)
        else:
            cfg = sample_config_for_fn(fn)
            if cfg is None:
                attempts += 1
                if attempts >= MAX_ATTEMPTS_PER_MODEL:
                    reason = "skipped_unavailable"
                    break
                continue
        raw_lag = rng.choice(RAW_MAX_LAG_CHOICES)
        attempts += 1
        wr: list = []
        q, acc, stab, rt, npt = run_forecast_one_config(
            df_model39, RESIST_IDS, TEST_YEAR, TEST_MONTHS, cfg, "engineered_features", raw_lag, wr
        )
        forecasts += 1
        improved = q > best_q + MIN_IMPROVEMENT_DELTA
        if improved:
            best_q = q
            best_params = dict(cfg)
            no_gain = 0
        else:
            no_gain += 1
        for fs in ("engineered_features", "raw_sequence_features"):
            attempt_log.append({
                "model_family": fn,
                "attempt_id": attempts,
                "params": json.dumps(cfg, default=str),
                "feature_set": fs,
                "quality": q,
                "accuracy": acc,
                "stability": stab,
                "runtime_sec": rt,
                "improved_flag": improved,
                "no_gain_streak": no_gain,
                "raw_max_lag": raw_lag,
            })
        trial_qualities.append((q, dict(cfg)))
        if no_gain >= EARLY_STOP_NO_GAIN:
            reason = "no_gain_10"
            break
    if not reason:
        reason = "max_attempts"
    family_summaries.append({
        "model_family": fn,
        "feature_set": "both_logged",
        "total_attempts": attempts,
        "total_forecasts": forecasts,
        "best_params": json.dumps(best_params, default=str) if best_params else "",
        "best_quality": best_q,
        "reason_for_stop": reason,
    })


def run_search_supervised(fn: str, feature_set: str):
    if sample_config_for_fn(fn) is None:
        family_summaries.append({
            "model_family": fn,
            "feature_set": feature_set,
            "total_attempts": 0,
            "total_forecasts": 0,
            "best_params": "",
            "best_quality": -1.0,
            "reason_for_stop": "skipped_unavailable",
        })
        return
    best_q = -1.0
    no_gain = 0
    attempts = 0
    forecasts = 0
    best_params = None
    reason = ""
    trial_qualities: list[tuple[float, dict]] = []
    refinement_queue: list[dict] = []
    refine_seeded = False
    while attempts < MAX_ATTEMPTS_PER_MODEL:
        if (not refine_seeded) and USE_REFINEMENT and attempts >= MAX_ATTEMPTS_PER_MODEL // 2 and trial_qualities:
            trial_qualities.sort(key=lambda x: -x[0])
            for _, cfg in trial_qualities[:3]:
                refinement_queue.extend([refine_config(cfg)])
            refine_seeded = True
        if refinement_queue:
            cfg = refinement_queue.pop(0)
        else:
            cfg = sample_config_for_fn(fn)
            if cfg is None:
                attempts += 1
                if attempts >= MAX_ATTEMPTS_PER_MODEL:
                    reason = "skipped_unavailable"
                    break
                continue
        raw_lag = rng.choice(RAW_MAX_LAG_CHOICES)
        attempts += 1
        wr = []
        q, acc, stab, rt, npt = run_forecast_one_config(
            df_model39, RESIST_IDS, TEST_YEAR, TEST_MONTHS, cfg, feature_set, raw_lag, wr
        )
        forecasts += 1
        improved = q > best_q + MIN_IMPROVEMENT_DELTA
        if improved:
            best_q = q
            best_params = dict(cfg)
            no_gain = 0
        else:
            no_gain += 1
        attempt_log.append({
            "model_family": fn,
            "attempt_id": attempts,
            "params": json.dumps(cfg, default=str),
            "feature_set": feature_set,
            "quality": q,
            "accuracy": acc,
            "stability": stab,
            "runtime_sec": rt,
            "improved_flag": improved,
            "no_gain_streak": no_gain,
            "raw_max_lag": raw_lag,
        })
        trial_qualities.append((q, dict(cfg)))
        if no_gain >= EARLY_STOP_NO_GAIN:
            reason = "no_gain_10"
            break
    if not reason:
        reason = "max_attempts"
    family_summaries.append({
        "model_family": fn,
        "feature_set": feature_set,
        "total_attempts": attempts,
        "total_forecasts": forecasts,
        "best_params": json.dumps(best_params, default=str) if best_params else "",
        "best_quality": best_q,
        "reason_for_stop": reason,
    })


for _fn in FN_ORDER:
    if _fn in UNIVARIATE_FNS:
        run_search_univariate(_fn)
    elif _fn in SUPERVISED_FNS:
        for _fs in ("engineered_features", "raw_sequence_features"):
            run_search_supervised(_fn, _fs)

df_attempts = pd.DataFrame(attempt_log)
df_family_summary = pd.DataFrame(family_summaries)
print("Попыток в логе:", len(df_attempts))
df_family_summary
```

```python
# --- Итоговые таблицы (совместимо с 30) ---
results = []
for _, row in df_attempts.iterrows():
    for rid in RESIST_IDS:
        results.append({
            "resist_id": rid,
            "Конфигурация": f"{row['model_family']}|{row['attempt_id']}",
            "Набор признаков": row["feature_set"],
            "Тип": "per-series-search",
            "Точность, %": row["accuracy"],
            "Стабильность, %": row["stability"],
            "Качество, %": row["quality"],
            "Время, сек": row["runtime_sec"],
            "Точн. fact_predict, %": np.nan,
            "Стаб. fact_predict, %": np.nan,
            "Кач. fact_predict, %": np.nan,
        })
df_results_full = pd.DataFrame(results)
df_best_per_family_fs = (
    df_attempts.loc[df_attempts.groupby(["model_family", "feature_set"])["quality"].idxmax()]
    .reset_index(drop=True)
)
print("Лучшие качества по семейству и feature_set:")
for fs in ["engineered_features", "raw_sequence_features"]:
    sub = df_attempts[df_attempts["feature_set"] == fs]
    if sub.empty:
        continue
    print(f"\n--- {fs} ---")
    print(sub.loc[sub.groupby("model_family")["quality"].idxmax()][["model_family", "quality", "params"]].to_string())

df_results = df_attempts.copy()
```

```python
# --- Excel (как в 30: Итог, Данные теста, По операциям по resist_id + лучшие конфиги) ---
from openpyxl import load_workbook

out_path = OUT_XLSX
out_path.parent.mkdir(parents=True, exist_ok=True)

WARN_COLS = [
    "resist_id", "operation_id", "model_kind", "model_label", "phase", "warning_category",
    "message", "interpretation", "n_train_points", "statsmodels_module",
]

def _short_sheet(name: str, max_len: int = 31) -> str:
    """Имя листа Excel (Excel sheet) — ограничение openpyxl."""
    return name[:max_len] if len(name) > max_len else name


def _cfg_from_attempt_row(row) -> dict:
    return json.loads(row["params"])


warning_rows_export: list = []

# Лучший глобальный прогон (для листов «Данные теста» / «По операциям» как в 30)
_best_sort = df_attempts.sort_values("quality", ascending=False)
if len(_best_sort) == 0:
    print("df_attempts пуст — детальные листы по факту не строим.")
    per_rid_best_global = {}
    per_rid_eng = {}
    per_rid_raw = {}
else:
    br = _best_sort.iloc[0]
    _cfg_g = _cfg_from_attempt_row(br)
    _combo_g = f"{br['model_family']}|{str(br['feature_set'])[:12]}"
    per_rid_best_global = run_forecast_collect_detail(
        df_model39, RESIST_IDS, TEST_YEAR, TEST_MONTHS,
        _cfg_g, br["feature_set"], int(br["raw_max_lag"]), _combo_g, warning_rows_export,
    )
    sub_e = df_attempts[df_attempts["feature_set"] == "engineered_features"]
    if len(sub_e) > 0:
        er = sub_e.sort_values("quality", ascending=False).iloc[0]
        per_rid_eng = run_forecast_collect_detail(
            df_model39, RESIST_IDS, TEST_YEAR, TEST_MONTHS,
            _cfg_from_attempt_row(er), "engineered_features", int(er["raw_max_lag"]),
            f"{er['model_family']}|eng", warning_rows_export,
        )
    else:
        per_rid_eng = {}
    sub_r = df_attempts[df_attempts["feature_set"] == "raw_sequence_features"]
    if len(sub_r) > 0:
        rr = sub_r.sort_values("quality", ascending=False).iloc[0]
        per_rid_raw = run_forecast_collect_detail(
            df_model39, RESIST_IDS, TEST_YEAR, TEST_MONTHS,
            _cfg_from_attempt_row(rr), "raw_sequence_features", int(rr["raw_max_lag"]),
            f"{rr['model_family']}|raw", warning_rows_export,
        )
    else:
        per_rid_raw = {}

sheet_triples = []
for rid in RESIST_IDS:
    df_itog = df_attempts.sort_values("quality", ascending=False).head(EXCEL_ITOG_TOP)
    if rid in per_rid_best_global:
        b = per_rid_best_global[rid]
        df_warn = pd.DataFrame([r for r in warning_rows_export if r.get("resist_id") == rid])
        if df_warn.empty:
            df_warn = pd.DataFrame(columns=WARN_COLS)
        else:
            df_warn = df_warn.reindex(columns=WARN_COLS)
        sheet_triples.append((rid, df_itog, b["df_detail"], b["df_metrics_op"], df_warn))
    else:
        df_empty = pd.DataFrame()
        sheet_triples.append((rid, df_itog, df_empty, df_empty, pd.DataFrame(columns=WARN_COLS)))

names_to_remove = []
for rid, _, _, _, _ in sheet_triples:
    names_to_remove.extend([
        _short_sheet(f"{rid} Итог"),
        _short_sheet(f"{rid} Данные теста"),
        _short_sheet(f"{rid} По операциям"),
        _short_sheet(f"{rid} Данные eng"),
        _short_sheet(f"{rid} Опер eng"),
        _short_sheet(f"{rid} Данные raw"),
        _short_sheet(f"{rid} Опер raw"),
    ])
    if RECORD_MODEL_WARNINGS:
        names_to_remove.append(_short_sheet(f"{rid} Предупреждения"))

base_sheets = ["Попытки", "Сводка_семейства", "Лучшие_по_семейству", "Итог_топ", "Лучшие_eng_raw"]
names_to_remove.extend(base_sheets)

if out_path.exists():
    wb = load_workbook(out_path)
    for sn in names_to_remove:
        if sn in wb.sheetnames:
            wb.remove(wb[sn])
    if len(wb.sheetnames) == 0:
        wb.create_sheet("_tmp")
    wb.save(out_path)
    wb.close()
    writer_mode = "a"
else:
    writer_mode = "w"

with pd.ExcelWriter(out_path, engine="openpyxl", mode=writer_mode) as w:
    df_attempts.to_excel(w, sheet_name="Попытки", index=False)
    df_family_summary.to_excel(w, sheet_name="Сводка_семейства", index=False)
    df_best_per_family_fs.to_excel(w, sheet_name="Лучшие_по_семейству", index=False)
    df_attempts.sort_values("quality", ascending=False).head(EXCEL_ITOG_TOP).to_excel(
        w, sheet_name="Итог_топ", index=False
    )
    # Сводка лучших eng vs raw (две строки-эталона)
    _rows = []
    if len(df_attempts[df_attempts["feature_set"] == "engineered_features"]) > 0:
        er = df_attempts[df_attempts["feature_set"] == "engineered_features"].sort_values("quality", ascending=False).iloc[0]
        _rows.append(er.to_dict())
    if len(df_attempts[df_attempts["feature_set"] == "raw_sequence_features"]) > 0:
        rr = df_attempts[df_attempts["feature_set"] == "raw_sequence_features"].sort_values("quality", ascending=False).iloc[0]
        _rows.append(rr.to_dict())
    if _rows:
        pd.DataFrame(_rows).to_excel(w, sheet_name="Лучшие_eng_raw", index=False)

    for rid, df_itog, df_detail, df_metrics_op, df_warn in sheet_triples:
        df_itog.to_excel(w, sheet_name=_short_sheet(f"{rid} Итог"), index=False)
        if not df_detail.empty:
            df_detail.to_excel(w, sheet_name=_short_sheet(f"{rid} Данные теста"), index=False)
        if not df_metrics_op.empty:
            df_metrics_op.to_excel(w, sheet_name=_short_sheet(f"{rid} По операциям"), index=False)
        if rid in per_rid_eng and per_rid_eng[rid]["df_detail"].shape[0] > 0:
            per_rid_eng[rid]["df_detail"].to_excel(w, sheet_name=_short_sheet(f"{rid} Данные eng"), index=False)
            per_rid_eng[rid]["df_metrics_op"].to_excel(w, sheet_name=_short_sheet(f"{rid} Опер eng"), index=False)
        if rid in per_rid_raw and per_rid_raw[rid]["df_detail"].shape[0] > 0:
            per_rid_raw[rid]["df_detail"].to_excel(w, sheet_name=_short_sheet(f"{rid} Данные raw"), index=False)
            per_rid_raw[rid]["df_metrics_op"].to_excel(w, sheet_name=_short_sheet(f"{rid} Опер raw"), index=False)
        if RECORD_MODEL_WARNINGS and not df_warn.empty:
            df_warn.to_excel(w, sheet_name=_short_sheet(f"{rid} Предупреждения"), index=False)

wb = load_workbook(out_path)
if "_tmp" in wb.sheetnames and len(wb.sheetnames) > 1:
    wb.remove(wb["_tmp"])
    wb.save(out_path)
wb.close()

print(f"Excel: {out_path.resolve()}")
for rid, df_itog, _, _, df_warn in sheet_triples:
    msg = (
        f"  «{_short_sheet(str(rid) + ' Итог')}» ({len(df_itog)} строк), "
        f"«Данные теста», «По операциям»; при наличии — eng/raw детализация"
    )
    if RECORD_MODEL_WARNINGS:
        msg += f", предупреждений: {len(df_warn)}"
    print(msg)
```
